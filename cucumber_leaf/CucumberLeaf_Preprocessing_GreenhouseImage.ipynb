# Practice changing background <br> with a random sample image

```python
import pixellib
from pixellib.tune_bg import alter_bg

change_bg = alter_bg()
change_bg.load_pascalvoc_model("deeplabv3_xception_tf_dim_ordering_tf_kernels.h5")
```

```python
# Chnage the background color to green

change_bg.color_bg(image_path = "sample.jpg",
                   colors = (0,128,0), 
                   output_image_name="green_bg.jpg")
```

```python
# Chnage the background img to flower background

change_bg.change_bg_img(f_image_path = "sample.jpg",
                        b_image_path = "flower_background.jpg", 
                        output_image_name="new_image.jpg")
```

```python
# Chnage the background color to black

change_bg.color_bg(image_path = "sample.jpg",
                   colors = (0,0,0), 
                   output_image_name="black_bg.jpg")
```

```python
import matplotlib.image as mpimg
import matplotlib.pyplot as plt

image = mpimg.imread('black_bg.jpg')
plt.imshow(image)
```
